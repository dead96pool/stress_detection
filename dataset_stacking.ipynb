```python
import pandas as pd
import sys
import os
```

```python
# simple stacking function

def simple_stack(stack_rows, df):
    df = df.copy()
    df.drop(columns = [0], inplace = True)

    rows_to_remove = df.shape[0] % stack_rows
    new_row_count = df.shape[0] // stack_rows

    df.reset_index(drop = True, inplace = True)
    df.drop(df.tail(rows_to_remove).index, inplace = True)

    #print("df shape:", df.shape)
    new_columns = []
    for i in range(stack_rows):
        new_columns.extend([f"{col}_{i}" for col in df.columns])


    stack_df = pd.DataFrame(columns = new_columns)

    #print(len(stack_df.columns))
    #print(rows_to_remove)
    #print(new_row_count)

    for i in range(new_row_count):
        #print(f"block = {i*stack_rows} to {(i+1) * stack_rows}")
        block = df.iloc[i*stack_rows: (i+1) * stack_rows].values.flatten()
        #print("block len =", len(block))
        stack_df.loc[i] = block

    return stack_df
```

```python
def extract_ids(filename):
    parts = filename.split("_")

    participant_id = int(parts[1])
    test_id = int(parts[2].split(".")[0])

    return participant_id, test_id
```

```python
def progress_bar(i, stress_files, stress_file, bar_len = 50, fill='█'):

    progress = ((i+1) / len(stress_files) ) * 100

    bar_fill = int(progress // 2)
    
    bar = bar_fill * fill + "-" * (bar_len - bar_fill)

    print(f"\rReading: {stress_file} |{bar}| {progress:.1f}%", end = "")
```

```python
stress_path = fr"datasets\2019_stress_detection_dataset\STRESS"
calm_path = fr"datasets\2019_stress_detection_dataset\CALM"
```

```python
calm_files = os.listdir(calm_path)
calm_files = sorted(calm_files, key = extract_ids)

stress_files = os.listdir(stress_path)
stress_files = sorted(stress_files, key = extract_ids)
```

```python
wanted_columns = [0]
wanted_columns.extend([i for i in range(7, 43)])
```

```python
# total calm data
# Initialize the DataFrame
calm_df = pd.DataFrame()


for i, calm_file in enumerate(calm_files):
    progress_bar(i, calm_files, calm_file)
   
    temp = pd.read_csv(os.path.join(calm_path, calm_file), sep=";", header=None)
    temp = temp[wanted_columns]
    calm_df = pd.concat([calm_df, temp], axis=0)
   
calm_df
```

```python
# total stress data
# Initialize the DataFrame
stress_df = pd.DataFrame()


for i, stress_file in enumerate(stress_files):
    progress_bar(i, stress_files, stress_file)
   
    temp = pd.read_csv(os.path.join(stress_path, stress_file), sep=";", header=None)
    temp = temp[wanted_columns]
    stress_df = pd.concat([stress_df, temp], axis=0)
   
stress_df
```
