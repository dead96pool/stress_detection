```python
import pandas as pd
import numpy as np
import os

from sklearn.preprocessing import LabelEncoder
```

```python
calm_path = r"2019_stress_detection_dataset\\CALM\\"
stress_path = r"2019_stress_detection_dataset\\STRESS\\"
```

## calm data

```python
# calm files
calm_df = pd.DataFrame()

for i in range(1,111):
    for j in range(1, 8):
        file_path = calm_path + f"CALM_{i}_{j}.csv"
        try:
            if os.path.exists(file_path):
                df = pd.read_csv(file_path, sep=";",header=None)
                calm_df = pd.concat([calm_df, df], axis=0)
            else:
                print(f"File {file_path} does not exist.")
        except FileNotFoundError:
            print(f"File {file_path} not found.")
```

```python
display(calm_df.isna().sum())
```

```python
#calm_df_filtered = calm_df.iloc[:,2:].dropna(axis = 1)
calm_df_filtered = calm_df.dropna(axis = 1)

calm_df_filtered.drop(columns= [0], inplace = True)

display(calm_df_filtered)
```

```python
calm_df_filtered.drop_duplicates(inplace = True)
display(calm_df_filtered)
```

```python
label_enc = LabelEncoder()
calm_df_filtered[2] = label_enc.fit_transform(calm_df_filtered[2]).astype("int64")
```

```python
print("Classes and their corresponding codes:")
for class_, code in zip(label_enc.classes_, range(len(label_enc.classes_))):
    print(f"{class_}: {code}")
```

```python
# combining rows to reduce the rows
# combining 100 rows so 5 secs worth data in a single row
def combining_rows(df, row_count):
    #row_count = 100  # Change this value to control how many rows to stack horizontally

    # Calculate the number of rows to add
    num_rows_to_add = row_count - (len(df) % row_count)

    # Generate NaN values for the new rows
    new_rows_data = pd.DataFrame({col: [np.nan] * num_rows_to_add for col in df.columns})

    # Append the new rows to the DataFrame
    # this is to make the rows in the dataframe a multiple of the row_count
    calm_combine_df = pd.concat([df, new_rows_data], ignore_index=True)

    n = len(calm_combine_df)
    # Calculate the number of resulting columns
    num_columns = calm_combine_df.shape[1] * row_count

    # Reshape the DataFrame into desired shape
    reshaped_values = np.reshape(calm_combine_df.values, (-1, row_count, calm_combine_df.shape[1]))
    column_names = [f"{col}_{i}" for i in range(row_count) for col in calm_combine_df.columns]
    calm_combine_df = pd.DataFrame(reshaped_values.reshape(-1, num_columns), columns=column_names)

    return calm_combine_df

# (df, rowcount)
calm_combine_df = combining_rows(calm_df_filtered, 20) # 1 secs  1*20
#calm_combine_df = combining_rows(calm_df_filtered, 60) # 3 secs  3*20
#calm_combine_df = combining_rows(calm_df_filtered, 100) # 5 secs 5*20
#calm_combine_df = combining_rows(calm_df_filtered, 140) # 7 secs 7*20
```

```python
#calm_combine_df.dropna(inplace = True)
#calm_combine_df["target"] = 0
#display(calm_combine_df)


calm_df_filtered["target"] = 0
display(calm_df_filtered)
```
